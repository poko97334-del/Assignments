# Q4. numbers.txt banayein, 1-5 tak numbers line-by-line likhein

```python
print("\nQ4:")
f = open("numbers.txt", "w")
for i in range(1, 6):
    f.write(str(i) + "\n")
f.close()
print("numbers.txt ban gayi.")
```

# Q5. numbers.txt read karke sum calculate karein

```python
print("\nQ5:")
f = open("numbers.txt", "r")
lines = f.readlines()
total = 0
for line in lines:
    total += int(line.strip())
f.close()
print("Sum:", total)
```

# Q9. source.txt ka data destination.txt mein copy karein

```python
print("\nQ9:")
with open("source.txt", "w") as f:
    f.write("Ye source file ka data hai.")

with open("source.txt", "r") as src:
    data = src.read()

with open("destination.txt", "w") as dest:
    dest.write(data)
print("Copy ho gaya.")
```

# Q14. IndexError handle karein

```python
print("\nQ14:")
my_list = [10, 20, 30]
try:
    print(my_list[5])
except IndexError:
    print("Error: Ye index list mein exist nahi karta.")
```

# Q15. KeyError handle karein


```python
print("\nQ15:")
student = {"name": "Ali"}
try:
    print(student["age"])
except KeyError:
    print("Error: 'age' key dictionary mein maujood nahi hai.")
```

# Q16. Multiple exceptions ek sath handle karein

```python
print("\nQ16:")
try:
    x = 10 / 0          # isay try karein, ya phir int("abc") try karein
except ZeroDivisionError:
    print("Error: Division by zero hua.")
except ValueError:
    print("Error: Invalid value di gayi hai.")
```

# Q17. finally block jo hamesha chale

```python
print("\nQ17:")
try:
    x = 10 / 2
    print("Result:", x)
except ZeroDivisionError:
    print("Error: Division by zero.")
finally:
    print("Ye finally block hai, hamesha chalta hai.")
```

# Q18. else block jo sirf tab chale jab error na aaye

```python
print("\nQ18:")
try:
    x = 10 / 2
except ZeroDivisionError:
    print("Error: Division by zero.")
else:
    print("Koi error nahi aaya. Result:", x)
```

# Q19. raise keyword se age validation

```python
print("\nQ19:")
def check_age(age):
    if age < 18:
        raise ValueError("Age must be 18 or older")
    else:
        print("Age valid hai:", age)

try:
    check_age(15)
except ValueError as e:
    print("Error:", e)
```

# Q20. File handling + Exception handling combine karein

```python
print("\nQ20:")
try:
    f = open("kisi_bhi_file.txt", "r")
    content = f.read()
    print(content)
except FileNotFoundError:
    print("Error: File nahi mili.")
finally:
    try:
        f.close()
        print("File close ho gayi.")
    except NameError:
        print("File open hi nahi hui thi, close karne ki zaroorat nahi.")
```
