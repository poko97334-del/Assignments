## **Q1**: Student class: private `name` aur `roll_number`, getter/setter methods.



```python
class Student:
    def __init__(self, name, roll_number):
        self.__name = name
        self.__roll_number = roll_number

    def get_name(self):
        return self.__name

    def set_name(self, name):
        self.__name = name

    def get_roll_number(self):
        return self.__roll_number

    def set_roll_number(self, roll_number):
        self.__roll_number = roll_number

s = Student("Ali", 101)
print(s.get_name(), s.get_roll_number())
s.set_name("Ahmed")
s.set_roll_number(202)
print(s.get_name(), s.get_roll_number())
```

# **Q2**: BankAccount: private `__balance`, deposit() aur withdraw() (balance zero se kam na ho).

```python
class BankAccount:
    def __init__(self, balance=0):
        self.__balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"Deposited {amount}. Balance: {self.__balance}")
        else:
            print("Invalid deposit amount")

    def withdraw(self, amount):
        if amount > self.__balance:
            print("Insufficient balance! Balance zero se kam nahi ho sakta.")
        else:
            self.__balance -= amount
            print(f"Withdrawn {amount}. Balance: {self.__balance}")

acc = BankAccount(1000)
acc.deposit(500)
acc.withdraw(300)
acc.withdraw(5000)
```

# **Q3**: Employee: private salary, negative salary par 'Invalid Salary'.

```python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.__salary = 0
        self.set_salary(salary)

    def get_salary(self):
        return self.__salary

    def set_salary(self, salary):
        if salary < 0:
            print("Invalid Salary")
        else:
            self.__salary = salary

e = Employee("Sara", 50000)
print(e.get_salary())
e.set_salary(-100)      # Invalid Salary
print(e.get_salary())   # purani salary hi rahegi
```

# **Q4**: Protected (`_`) vs Private (`__`) variable ka farq.

```python
class Demo:
    def __init__(self):
        self._protected = "Main protected hoon"   # convention: sirf class/subclass use kare
        self.__private = "Main private hoon"      # name mangling hoti hai

class Child(Demo):
    def show(self):
        print(self._protected)   # subclass mein access ho jata hai

d = Demo()
print(d._protected)            # access ho jata hai (sirf convention hai)
# print(d.__private)           # AttributeError aayega
print(d._Demo__private)        # name mangling ki wajah se aise access hota hai
Child().show()

# Farq: _var = sirf warning/convention (access possible)
#       __var = Python naam badal deta hai (_ClassName__var), direct access nahi hota
```

# **Q5**: Encapsulation ka faida aur private variable direct access par error.

# **Q9**: `super()` ka use child constructor mein.

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

class Student(Person):
    def __init__(self, name, age, course):
        super().__init__(name, age)   # parent ka constructor call
        self.course = course

    def show(self):
        print(f"Name: {self.name}, Age: {self.age}, Course: {self.course}")

Student("Ali", 20, "Python").show()
```

# **Q10**: Hierarchical Inheritance: Shape -> Circle, Square.

```python
import math

class Shape:
    def area(self):
        print("Area calculate hoga")

class Circle(Shape):
    def __init__(self, r):
        self.r = r
    def area(self):
        return math.pi * self.r ** 2

class Square(Shape):
    def __init__(self, side):
        self.side = side
    def area(self):
        return self.side ** 2

print("Circle area:", round(Circle(5).area(), 2))
print("Square area:", Square(4).area())
```

# **Q12**: Polymorphism via function `make_sound(animal_object)`.

# **Q13**: Operator Overloading with `__add__`.

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Point(self.x + other.x, self.y + other.y)

    def __str__(self):
        return f"Point({self.x}, {self.y})"

p1 = Point(2, 3)
p2 = Point(4, 5)
print(p1 + p2)
```

# **Q15**: Method Overloading Python mein support hoti hai ya nahi, aur alternative.


```python
class Calc:
    def add(self, a, b):
        return a + b
    def add(self, a, b, c):   # ye pehle wale ko replace kar dega
        return a + b + c

try:
    Calc().add(1, 2)
except TypeError as e:
    print("Error:", e)

# Solution 1: default arguments
class Calc2:
    def add(self, a, b, c=0):
        return a + b + c

print(Calc2().add(1, 2))
print(Calc2().add(1, 2, 3))

# Solution 2: *args
class Calc3:
    def add(self, *args):
        return sum(args)

print(Calc3().add(1, 2))
print(Calc3().add(1, 2, 3, 4))

```

# **Q17**: Laptop aur Desktop mein `process()` implement karna.

```python
from abc import ABC, abstractmethod

class Computer(ABC):
    @abstractmethod
    def process(self):
        pass

class Laptop(Computer):
    def process(self):
        print("Laptop: battery-efficient processing")

class Desktop(Computer):
    def process(self):
        print("Desktop: high-performance processing")

Laptop().process()
Desktop().process()
```

# **Q18**: Abstract class Car: concrete + abstract method.

```python
class Car(ABC):
    def start_engine(self):          # concrete method
        print("Engine started")

    @abstractmethod
    def drive(self):                 # abstract method
        pass

class Sedan(Car):
    def drive(self):
        print("Sedan is driving smoothly")

s = Sedan()
s.start_engine()
s.drive()
```

# **Q20**: Payment abstract class, CreditCardPayment aur UPIPayment.

```python
from abc import ABC, abstractmethod

class Payment(ABC):
    @abstractmethod
    def pay(self, amount):
        pass

class CreditCardPayment(Payment):
    def pay(self, amount):
        print(f"Rs. {amount} Credit Card se pay kiye gaye")

class UPIPayment(Payment):
    def pay(self, amount):
        print(f"Rs. {amount} UPI se pay kiye gaye")

for method in (CreditCardPayment(), UPIPayment()):
    method.pay(1500)
```
