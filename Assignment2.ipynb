#Q6 - [2,-5,10,-8,0] POSITIVE/NEGATIVE/ZERO

```python
nums = [2, -5, 10, -8, 0]
for n in nums:
    if n > 0:
        print("Positive")
    elif n < 0:
        print("Negative")
    else:
        print("Zero")
```

#Q15 - COUNT 'apple'

```python
count = 0
fruits = ["apple", "mango", "apple", "banana", "apple"]
for f in fruits:
    if f == "apple":
        count += 1
print(count)
```
