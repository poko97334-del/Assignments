# Q1. 'Hello'

```python
print("Q1:")
count = 0
while count < 3:
    print("Hello")
    count += 1
```

# Q2. 10 to 1  (reverse counting)

```python
print("\nQ2:")
n = 10
while n >= 1:
    print(n)
    n -= 1
```

# Q3. 1 to 10  (Print odd numbers)

```python
print("\nQ3:")
i = 1
while i <= 10:
    if i % 2 != 0:
        print(i)
    i += 1
```

# Q4. 5 ka table sirf 5 tak print karein


```python
print("\nQ4:")
i = 1
while i <= 5:
    print(f"5 x {i} = {5*i}")
    i += 1
```

# Q9. "Ali" ke har letter ko index se print karein

```python
print("\nQ9:")
word = "Ali"
i = 0
while i < len(word):
    print(word[i])
    i += 1
```

# Q10. Print from 1 to 5, skip to count 3

```python
print("\nQ10:")
i = 0
while i < 5:
    i += 1
    if i == 3:
        continue
    print(i)
```

# ---------- PART 2: FUNCTIONS ----------

# Q1. greet(name) -> "Hello" kahe


```python
def greet(name):
    print(f"Hello, {name}")

greet("Ahmed")
```

# Q2. add_numbers(a, b) -> sum return kare

```python
def add_numbers(a, b):
    return a + b

print(add_numbers(5, 7))
```

# Q3. check_even_odd(n) -> even ya odd bataye

```python
def check_even_odd(n):
    if n % 2 == 0:
        return "Even"
    else:
        return "Odd"

print(check_even_odd(9))
```

# Q4. find_maximum(a, b, c) -> sabse bada number

```python
def find_maximum(a, b, c):
    return max(a, b, c)

print(find_maximum(3, 9, 6))
```

# Q5. circle_area(radius) -> area calculate kare

```python
def circle_area(radius):
    pi = 3.14159
    return pi * radius * radius

print(circle_area(4))
```

# Q6. celsius_to_fahrenheit(c) -> temperature convert kare

```python
def celsius_to_fahrenheit(c):
    return (c * 9/5) + 32

print(celsius_to_fahrenheit(25))
```

# Q7. is_palindrome(word) -> palindrome check kare

```python
def is_palindrome(word):
    return word == word[::-1]

print(is_palindrome("madam"))
```

# Q8. get_square(n) -> number ka square return kare

```python
def get_square(n):
    return n * n

print(get_square(6))
```

# Q9. calculate_factorial(n) -> factorial return kare

```python
def calculate_factorial(n):
    result = 1
    i = 1
    while i <= n:
        result *= i
        i += 1
    return result

print(calculate_factorial(5))
```

# Q10. count_vowels(text) -> vowels ki tadad bataye

```python
def count_vowels(text):
    vowels = "aeiouAEIOU"
    count = 0
    for ch in text:
        if ch in vowels:
            count += 1
    return count

print(count_vowels("Programming"))
```
